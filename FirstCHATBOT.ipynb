```python



def chatbot():

    print("================================")
    print("       BIENVENUE AU CHATBOT")
    print("================================")
    print("Chatbot : Bonjour ! Comment puis-je t'aider ?")
    print("Tape 'quitter' pour terminer la conversation.")

    while True:

        
        message = input("Vous : ")

        
        message = message.lower()

        
        if message == "bonjour" or message == "salut":
            print("Chatbot : Bonjour ! Comment vas-tu ?")

        elif message == "ça va":
            print("Chatbot : Je vais très bien, merci !")

        elif message == "comment tu t'appelles":
            print("Chatbot : Je suis un chatbot créé avec Python.")

        elif message == "que peux-tu faire":
            print("Chatbot : Je peux répondre à quelques questions simples.")

        elif message == "merci":
            print("Chatbot : Avec plaisir !")

        elif message == "quitter":
            print("Chatbot : Au revoir ! À bientôt.")
            break

        else:
            print("Chatbot : Désolé, je ne comprends pas ta question.")


chatbot()
```
